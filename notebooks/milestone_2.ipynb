```python
# Check with the first 5 attacks
import pandas as pd
import time
from pathlib import Path
from rag_pipeline import answer_question

attack_file = Path("../results/security_attack_suite.csv")
attack_df = pd.read_csv(attack_file)

# Test a small batch first
test_rows = attack_df.head(5).copy()
before_results = []

for _, row in test_rows.iterrows():
    print(f"Testing {row['attack_id']}: {row['attack_type']}")

    start = time.perf_counter()

    try:
        result = answer_question(
            row["payload"],
            pipeline["rag_chain"])

        elapsed = time.perf_counter() - start

        before_results.append({
            "attack_id": row["attack_id"],
            "before_result": str(result.get("result", "")),
            "before_blocked": result.get("blocked", False),
            "before_retrieval_relevant": result.get("retrieval_relevant", "N/A"),
            "before_answer_supported": result.get("answer_supported", "N/A"),
            "before_latency_seconds": round(elapsed, 3),
            "before_error": ""})

    except Exception as e:
        before_results.append({
            "attack_id": row["attack_id"],
            "before_result": "",
            "before_blocked": False,
            "before_retrieval_relevant": "N/A",
            "before_answer_supported": "N/A",
            "before_latency_seconds": None,
            "before_error": str(e)})

    print("  Blocked:", before_results[-1]["before_blocked"])
    print("  Latency:", before_results[-1]["before_latency_seconds"], "seconds")

before_df = pd.DataFrame(before_results)

# Merge the observed results into the attack suite
for col in before_df.columns:
    if col != "attack_id":
        attack_df.loc[
            attack_df["attack_id"].isin(before_df["attack_id"]), col
        ] = attack_df["attack_id"].map(
            before_df.set_index("attack_id")[col])

attack_df.to_csv(attack_file, index=False)

print(f"\nSaved observed results to {attack_file}")
```

## Compare M2 Golden Set results Before Defenses vs After
